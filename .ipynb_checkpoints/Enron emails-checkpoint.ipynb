```python
import pandas as pd


DATASET_PATH = "datasets/CC_1/emails.csv"  # or .json

if DATASET_PATH.endswith(".csv"):
    df = pd.read_csv(DATASET_PATH)
elif DATASET_PATH.endswith(".json"):
    df = pd.read_json(DATASET_PATH, lines=True)
else:
    raise ValueError("Unsupported format")

print("Dataset loaded successfully")
print("Shape:", df.shape)
```

```python
print("\n===== COLUMN INFO =====\n")
print(df.dtypes)

print("\n===== SAMPLE ROWS (SANITIZED) =====\n")
print(df.head(3))
```

```python
text_candidates = []

for col in df.columns:
    if df[col].dtype == "object":
        avg_len = df[col].astype(str).str.len().mean()
        if avg_len > 20:
            text_candidates.append((col, round(avg_len, 2)))

print("\n===== TEXT COLUMN CANDIDATES =====")
for col, avg_len in text_candidates:
    print(f"{col}: avg length = {avg_len}")
```

```python
# CHANGE THIS if you know the label column
POSSIBLE_LABELS = ["label", "class", "attack_type", "is_phishing"]

label_col = None
for c in POSSIBLE_LABELS:
    if c in df.columns:
        label_col = c
        break

if label_col:
    print(f"\n===== LABEL COLUMN: {label_col} =====")
    print(df[label_col].value_counts())
else:
    print("\n⚠️ No obvious label column detected")
```

```python
import numpy as np

def text_stats(col):
    lengths = df[col].astype(str).str.split().apply(len)
    return {
        "mean": round(lengths.mean(), 2),
        "median": int(lengths.median()),
        "max": int(lengths.max())
    }

print("\n===== TEXT LENGTH STATS =====")
for col, _ in text_candidates:
    print(col, text_stats(col))
```

```python
URGENCY_TERMS = [
    "urgent", "immediately", "asap", "action required", "final notice",
    "verify", "suspended", "compromised", "limited time", "now"
]

def urgency_ratio(col):
    text = df[col].astype(str).str.lower()
    hits = sum(text.str.contains(term).sum() for term in URGENCY_TERMS)
    return round(hits / len(df), 4)

print("\n===== URGENCY SIGNALS =====")
for col, _ in text_candidates:
    print(col, "urgency_ratio:", urgency_ratio(col))
```

```python
sender_cols = [c for c in df.columns if "sender" in c.lower() or "from" in c.lower()]
receiver_cols = [c for c in df.columns if "to" in c.lower() or "receiver" in c.lower()]

print("\n===== SENDER COLUMNS =====")
print(sender_cols)

print("\n===== RECEIVER COLUMNS =====")
print(receiver_cols)
```

```python
time_cols = [c for c in df.columns if "time" in c.lower() or "date" in c.lower()]

print("\n===== TIME COLUMNS =====")
print(time_cols)

if time_cols:
    col = time_cols[0]
    df[col] = pd.to_datetime(df[col], errors="coerce")
    print("\nMessages per hour:")
    print(df[col].dt.hour.value_counts().sort_index())
```

```python
print("\n===== MISSING VALUES (%) =====")
print((df.isnull().mean() * 100).round(2))
```

```python
#Step 10

import pandas as pd
from email import policy
from email.parser import BytesParser
from tqdm import tqdm
import re


def parse_email(raw_text):
    try:
        msg = BytesParser(policy=policy.default).parsebytes(
            raw_text.encode("utf-8", errors="ignore")
        )
        return {
            "from": msg.get("From"),
            "to": msg.get("To"),
            "cc": msg.get("Cc"),
            "subject": msg.get("Subject"),
            "date": msg.get("Date"),
            "message_id": msg.get("Message-ID"),
            "in_reply_to": msg.get("In-Reply-To"),
            "references": msg.get("References"),
            "body": msg.get_body(preferencelist=("plain",))
                    .get_content() if msg.get_body() else ""
        }
    except Exception:
        return None

parsed_rows = []

for text in tqdm(df["message"].iloc[:50000]):  # LIMIT for now
    parsed = parse_email(text)
    if parsed:
        parsed_rows.append(parsed)

parsed_df = pd.DataFrame(parsed_rows)
print(parsed_df.head())
print(parsed_df.isnull().mean())
```

```python
#Step 11
def is_external(email):
    if not email or "@" not in email:
        return None
    return not email.lower().endswith("@enron.com")

parsed_df["external_sender"] = parsed_df["from"].apply(is_external)
print(parsed_df["external_sender"].value_counts(dropna=False))
```

```python
#Step12
URGENCY_TERMS = [
    "urgent", "immediately", "asap", "action required",
    "verify", "suspended", "compromised", "final notice"
]

def urgency_score(text):
    if not isinstance(text, str):
        return 0
    text = text.lower()
    return sum(term in text for term in URGENCY_TERMS)

parsed_df["subject_urgency"] = parsed_df["subject"].apply(urgency_score)
parsed_df["body_urgency"] = parsed_df["body"].apply(urgency_score)

print(parsed_df[["subject_urgency", "body_urgency"]].describe())
```

```python
#Step 13
parsed_df["date"] = pd.to_datetime(
    parsed_df["date"],
    errors="coerce",
    utc=True
)

print(parsed_df["date"].dtype)
```

```python
parsed_df["hour"] = parsed_df["date"].dt.hour
print(parsed_df["hour"].value_counts().sort_index())
```

```python
import re

def clean_email(text):
    if not isinstance(text, str):
        return ""
    
    
    text = re.split(r"\nOn .* wrote:\n", text)[0]
    text = re.split(r"\n-----Original Message-----", text)[0]
    
    
    text = re.split(r"\n--\s*\n", text)[0]
    
    
    text = re.sub(r"\s+", " ", text).strip()
    
    return text

parsed_df["clean_body"] = parsed_df["body"].apply(clean_email)
parsed_df["body_len"] = parsed_df["clean_body"].str.split().apply(len)

parsed_df[["body_len"]].describe()
```

```python
parsed_df["subject_len"] = parsed_df["subject"].fillna("").str.split().apply(len)

parsed_df["empty_subject"] = parsed_df["subject"].isnull() | (parsed_df["subject"].str.strip() == "")

print(parsed_df["empty_subject"].value_counts(normalize=True))
print(parsed_df["subject_len"].describe())
```

```python
import networkx as nx

G = nx.DiGraph()

for _, row in parsed_df.iterrows():
    if row["from"] and row["to"]:
        sender = row["from"].lower()
        receivers = [r.strip().lower() for r in row["to"].split(",")]
        for r in receivers:
            G.add_edge(sender, r, weight=G.get_edge_data(sender, r, {"weight": 0})["weight"] + 1)

print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())
```

```python
degrees = [d for _, d in G.degree()]
print(pd.Series(degrees).describe())
```

```python
authority_scores = nx.out_degree_centrality(G)
parsed_df["sender_authority"] = parsed_df["from"].map(authority_scores)

parsed_df["sender_authority"].describe()
```

```python
hour_dist = parsed_df["hour"].value_counts(normalize=True).sort_index()
print(hour_dist)
```

```python
parsed_df = parsed_df.sort_values("date")

parsed_df["time_delta"] = parsed_df["date"].diff().dt.total_seconds()
parsed_df["time_delta"].describe()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    max_features=20000,
    stop_words="english",
    min_df=10
)

X = vectorizer.fit_transform(parsed_df["clean_body"])

print("TF-IDF shape:", X.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

sample_sim = cosine_similarity(X[:1000], X[:1000])
print("Mean similarity:", sample_sim.mean())
```

```python
from sklearn.ensemble import IsolationForest

iso = IsolationForest(
    n_estimators=200,
    contamination=0.05,
    random_state=42
)

subset_X = X[:5000].toarray()

subset_scores = iso.fit_predict(subset_X) 

parsed_df["anomaly_score"] = 0 

parsed_df.iloc[:5000, parsed_df.columns.get_loc("anomaly_score")] = subset_scores

print(parsed_df["anomaly_score"].head(10))
```

```python
from sklearn.ensemble import IsolationForest

batch_size = 5000
parsed_df["anomaly_score"] = 0

for start in range(0, X.shape[0], batch_size):
    end = min(start + batch_size, X.shape[0])
    batch_X = X[start:end].toarray()
    
    iso = IsolationForest(
        n_estimators=200,
        contamination=0.05,
        random_state=42
    )
    
    scores = iso.fit_predict(batch_X)
    parsed_df.iloc[start:end, parsed_df.columns.get_loc("anomaly_score")] = scores
    
    # Print progress
    print(f"Processed rows {start} to {end-1} | Anomalies in this batch: {(scores==-1).sum()}")

```

```python
parsed_df.to_parquet("enron_processed_baseline.parquet")
```
